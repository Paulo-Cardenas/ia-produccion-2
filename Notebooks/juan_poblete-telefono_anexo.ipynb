```python
import pandas as pd
```

```python
df = pd.read_excel("plan_de_compras_2025.xlsx")
```

```python
# #Transformacion titulos columnas a minusculas y sin espacios
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)

print(df.columns.tolist())
```

```python
# #Filtra ID Propyecto PC25
codigo_proceso = "PC25"

df = df[
    df["id proyecto"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
    .str.endswith(codigo_proceso.upper())
].copy()

print("Registros después del filtro:", len(df))
```

```python
#Limpia Telefonos
telefono_limpio = (
    df["teléfono responsable"]
    .fillna("")
    .astype(str)
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
    .str.strip()
)

df["anexo"] = telefono_limpio.str[-4:]

df["teléfono responsable"] = (
    telefono_limpio
    .str.slice(-9)
    .str.zfill(9)
)

df["teléfono responsable"] = (
    df["teléfono responsable"]
    .apply(
        lambda x: "2" + x[1:]
        if x.startswith("0") and x != ""
        else x
    )
)
```

```python
# #Solo para validar, despues se puede borrar
display(
    df[
        [
            "id proyecto",
            "teléfono responsable",
            "anexo"
        ]
    ].head(20)
)
```
