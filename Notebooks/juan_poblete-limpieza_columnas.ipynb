```python
import pandas as pd
```

```python
df = pd.read_excel("plan_de_compras_2025.xlsx")
```

```python
#Validacion Nombres y cantidades de columnas (conservar o eliminar)
print("Cantidad de filas:", len(df))
print("Cantidad de columnas:", len(df.columns))

print("\nColumnas:")
for columna in df.columns:
    print(repr(columna))
```

```python
#Transformacion titulos columnas a minusculas y sin espacios
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)

print(df.columns.tolist())
```

```python
#Filtra ID Propyecto PC25
codigo_proceso = "PC25"

df = df[
    df["id proyecto"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.upper()
    .str.endswith(codigo_proceso.upper())
].copy()

print("Registros después del filtro:", len(df))
```
